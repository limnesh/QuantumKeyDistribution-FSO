## 1. What the satellite nodes do

Each available optical link creates a shared pairwise key between its two endpoints.
Trusted satellite nodes can relay a ground-to-ground key with authenticated classical
messages, consuming one bit of link key on every hop for each delivered end-to-end bit.
**Every relay must be trusted.** Relays hold classical keys; this model has no entanglement
swapping, quantum memories or quantum repeaters. A satellite trusted-relay experiment
motivates the single-satellite case ([Liao et al., 2018](https://arxiv.org/abs/1801.04418)).
Inter-satellite trusted networks motivate extending that graph
([Vergoossen et al., 2020](https://arxiv.org/abs/1903.07845)).

Each candidate edge has its own optical terminals and full pulse-rate budget. Thus one
satellite can operate several links at the same time; adding nodes or edges can increase
total hardware and pulse resources. Shared keys are modeled as **undirected capacities**,
although ground-link optical pulses use the satellite-to-ground downlink budget.

At each time sample, one route is selected to maximize its weakest link:

$$K_{AB}(t)=\max_{P:A\rightarrow B}\min_{e\in P}K_e(t).$$

Only positive-key edges enter the routing graph. No route gives zero delivery rate.
The integral $\int K_{AB}(t)\,dt$ is a simultaneous-route **ideal asymptotic benchmark**,
assuming negligible relay/authentication overhead. It is not a generated key string,
a certified finite-key length, or a multi-route maximum flow. No cross-time inventory
scheduler is used for this route integral.

## 3. Optical links and QKD benchmark

A ground link is available only when its **actual** elevation is nonnegative
and meets the elevation cutoff. The inherited stage-2 downlink model applies
atmospheric attenuation only to a finite shell near Earth, while the Gaussian
beam spreads over the full slant range. Negative elevations are masked after
evaluating a clamped elevation; below-horizon links cannot create keys.

An inter-satellite link (ISL) is available only if its finite line segment clears
Earth plus the selected clearance. ISLs use their own divergence, receiver radius,
pointing loss and background yield, with **no atmospheric attenuation**. Their
constant same-orbit angular separations give constant link distances here.

For a path length $L$ in metres and receiver radius $a$,

$$w(L)=\sqrt{w_0^2+(\vartheta L)^2},\qquad
  \eta_g=1-\exp[-2a^2/w(L)^2].$$

Total detection efficiency includes aperture collection, pointing, optical and
detector efficiencies, plus ground-link atmospheric loss. With mean signal photon
number $\mu$, background yield $Y_0$, and optical error $e_d$,

$$S=1-e^{-\mu\eta},\quad Q_\mu=S+(1-S)Y_0,\quad
  E_\mu=\frac{e_dS+\tfrac12(1-S)Y_0}{Q_\mu}.$$

The stage-2 model supplies $Y_1=\eta+(1-\eta)Y_0$ and its corresponding $e_1$.
The per-edge model-known infinite-decoy benchmark is

$$K_e=f_p p_s q\max\{\mu e^{-\mu}Y_1[1-h_2(e_1)]
  -f_{EC}Q_\mu h_2(E_\mu),0\}.$$

This is the asymptotic decoy-state rate form
([Ma et al., 2005](https://arxiv.org/abs/quant-ph/0503005)), evaluated using
known model quantities rather than statistical decoy estimates. Unavailable
contacts have zero efficiency/rate and undefined QBER (`NaN`). Geometric
availability alone does not guarantee a positive secret-key benchmark.

## 6. Simultaneous relay versus later relay of stored keys

One satellite can establish keys with the two stations at different times
and relay a paired amount after both contacts. With independent terminals,
unlimited classical storage and negligible overhead, a post-window upper
bound is

$$N_{\rm stored}=\min\!\left\{\int K_{A,S}(t)dt,\int K_{S,B}(t)dt\right\}.$$

This can exceed $\int\min\{K_{A,S},K_{S,B}\}dt$, which requires simultaneous
positive capacity. It does not imply that the stored amount was available
earlier in the window, and it still relies on a trusted satellite.

The following larger-separation case has separate contacts within the
observation window. It illustrates positive stored-pair capacity even when
there is no simultaneous end-to-end route.

```python
separated = simulate_network(replace(p, ground_separation_km=4500))
table(["Scenario", "Simultaneous benchmark (Mbit)", "Stored-pair bound (Mbit)"], [
    [label, f'{scenario["metrics"]["integrated_key_bits"]/1e6:.4f}',
     f'{scenario["metrics"]["stored_pair_bits"]/1e6:.4f}']
    for label, scenario in [("Edited/default separation", result), ("4500 km separation", separated)]
])
fig, ax = plt.subplots(figsize=(10.5, 3.6), layout="constrained")
for edge in separated["edges"]:
    ax.plot(separated["time_s"]/60, edge["key_rate_bps"]/1000, label=edge["label"], linewidth=2)
ax.set(xlabel="Time from orbit midpoint (min)", ylabel="Link capacity (kbit/s)",
       title="Two separated ground contacts: keys can be stored for later relay")
ax.legend()
plt.show()
```

## 7. Mode B — Entanglement-Based Satellite QKD — BBM92

The satellite distributes a Bell pair, sending one photon to each ground receiver.
It does not measure or receive a final key in this model. Each receiver independently
chooses between two complementary polarization bases; matching-basis coincidences
are retained. Error estimation, reconciliation leakage and privacy amplification are
represented by asymptotic rate terms, not a finite measurement/LDPC transcript.

Using the already computed **total** efficiencies (including detector efficiency once),
the ideal true coincidence rate is $C_{true}=B\eta_A\eta_B$. Both ground links must
be visible at the same time. Singles on one arm cannot be stored for a later pair.

We use the high-loss CW model of [Neumann et al. (2021)](https://arxiv.org/abs/2103.14639),
equations 3, 6, 8, 10, 13–17:

$$S_i=B\eta_i+D_i+N_i,\quad
C_{acc}=\frac{(1-e^{-S_A\tau})(1-e^{-S_B\tau})}{\tau},\quad
g=\operatorname{erf}(\sqrt{\ln2}\,\tau/t_\Delta).$$

$\tau$ is the **full** coincidence window and $t_\Delta$ the relative timing FWHM;
timing delays are assumed corrected. $D_i,N_i$ are registered count rates summed
over a receiver's detectors, not per-gate probabilities. For low occupancy,
$C_{acc}\simeq S_A S_B\tau$. With identical independent per-arm error $e_d$ inherited
from the link model, $e_{pair}=2e_d(1-e_d)$:

$$C=gC_{true}+C_{acc},\quad
E=\frac{e_{pair}gC_{true}+C_{acc}/2}{C},\quad
K=qC\max[1-f_{EC}h_2(E)-h_2(E),0].$$

The last expression is an **asymptotic secret-key-rate estimate**, assuming symmetric
basis-independent errors so the phase-error estimate equals the bit error. It is
not the trusted-relay minimum-link formula. Zero coincidences give undefined QBER
and zero key; no common visibility gives zero coincidences and zero key.

Scope: high-loss channels, CW temporal statistics and low count occupancy. The code
rejects common-contact efficiencies above 0.1 or singles-times-window above 0.1,
a conservative declared applicability envelope, not a security threshold. Detector
saturation and detailed multipair/source physics are not modeled. No wavelength knob
is added because divergence/loss/detector inputs do not derive from wavelength.

**Security assumptions:** trusted basis-independent receivers, independent random
basis choices, tests in both bases, a valid qubit/squashing treatment of detection,
and authenticated classical communication are required. The numerical noise model
assumes rather than proves these conditions. Not assigning the satellite a key does
not establish arbitrary untrusted-source or device-independent security. No finite-key
composable proof, quantum memory, swapping or repeater is implemented.

## 8. Fair interpretation and a separate-contact experiment

The common optical conditions are the circular orbit, atmosphere, apertures,
divergence, pointing and detector efficiencies. Hardware resources differ:
Mode A has an independent weak-coherent-pulse clock per downlink (default 100 MHz,
signal allocation 0.8, mean 0.5 photons per signal pulse); Mode B uses a CW pair source
(10 million pairs/s) and two ground polarization analyzers. The default 200 registered
noise counts/s per BBM92 receiver is approximately 2e-6 noise probability in a
hypothetical 10 ns BB84 gate. The existing BB84 background yield is an independent
per-gate input, not automatically a count rate. Adjust count and gate assumptions
consistently for a calibrated comparison. No universal architecture ranking follows.

Trusted storage starts with empty classical pools. At each interval end, the model
credits generated link key amounts and serves unlimited demand, consuming one bit
from each pool per delivered bit. Thus cumulative delivered bits are the minimum
of the two cumulative generation curves. Leftover pools are recorded. These expected
budgets assume negligible relay/authentication delay and no expiration; they do not
implement finite-block secure key release. The simultaneous-route curve remains a
distinct benchmark. Stage 4 has no inventory scheduler.

```python
distant_result = simulate_network(replace(p, ground_separation_km=4000))
distant = compare_stage3(distant_result, bbm, "trusted")
table(["Separation (km)", "Common contact (s)", "Trusted stored relay (expected bit)", "BBM92 (estimated bit)"], [
    [item["network"]["parameters"].ground_separation_km,
     item["bbm92"]["metrics"]["common_visibility_s"],
     f'{item["trusted"]["delivered_bits"]:,.3f}',
     f'{item["bbm92"]["metrics"]["integrated_key_bits"]:,.3f}']
    for item in (comparison, distant)])
assert not distant["bbm92"]["common_visibility"].any()
if distant["bbm92"]["available"]:
    assert distant["bbm92"]["metrics"]["integrated_key_bits"] == 0
else:
    print("BBM92 estimate unavailable:", distant["bbm92"]["reason"])
inventory = distant["trusted"]["inventory"]
for side in ("a", "b"):
    np.testing.assert_allclose(inventory[f"generated_{side}_bits"],
                               inventory["delivered_bits"] + inventory[f"pool_{side}_bits"])
print("Separate contacts: stored classical keys can relay; entangled pairs cannot cross passes.")
```

## 10. Interpretation and limitations

- Geometry is an ideal static spherical Earth with coplanar circular satellites.
  Real coverage needs orbit ephemerides, ground coordinates and Earth rotation.
- Ground links have clear-sky effective-shell attenuation and a simple Gaussian
  beam budget. Clouds, weather, turbulence, scintillation, tracking acquisition,
  Doppler constraints and pointing jitter are not simulated.
- Independent per-edge optical terminals and pulse budgets allow simultaneous
  operation. No shared-terminal or energy scheduling constraints are imposed.
- Trusted relays expose the end-to-end key to participating satellite nodes.
  This stage has no quantum repeater, quantum memory or entanglement protocol.
- Model-known infinite-decoy asymptotic rates exclude statistical decoy bounds,
  finite-key penalties and authentication/relay costs. These are benchmarks,
  not certified secret keys or a simulated LDPC/hash transcript.
- The route benchmark chooses a single simultaneous route. It does not add
  multiple routes, optimize total flow or schedule stored-key inventories.
- Contacts and rates are sampled; connected duration and key integrals depend
  on time resolution. Increase `time_points` to examine convergence.

Validation uses this stage's local `test_network_model.py` and corresponding
Octave tests. A deterministic JSON fixture compares both implementations for
one, two, three and four satellites, plus disabled inter-satellite links.

## 11. DPS protocol statistics and atmospheric fading

This extension uses `shared/physics.py`, `shared/scenarios.py` and `shared/reporting.py`.
Earlier sections retain their historical deterministic models and finite BB84 demonstrations.
**DPS uses relative phases between coherent pulses, a one-pulse delay interferometer,
two threshold detectors and explicit valid-slot accounting.** It is not the earlier DPS teaching curve.

**DPS protocol QBER** is a detection statistic. **DPS illustrative rate proxy** uses a
collision-inspired privacy term from the restricted individual-attack analysis, combined
with the detector model described in `DPS_THEORY.md`. A DPS-specific composable secret-key
security bound has not been implemented or established. Its integral is an integrated
proxy, not certified secret-key material. The old BB84/Eve/LDPC/Toeplitz example is not DPS.

Fading is a normalized collected-power multiplier: `exp(sigma*X - sigma**2/2)`;
`sigma**2` is log-irradiance variance. Temporal correlation is phenomenological,
with `rho=exp(-dt/tau_c)` and stationary initialization. The detector factor is fixed;
optical clipping is reported. Ground links have independent random streams; vacuum
inter-satellite links do not receive atmospheric fading. No measured data are used.

Change the protocol and mode below, then restart and run all. Defaults here explicitly
select correlated fading for comparison; library defaults remain `none`.
Stage 3 retains trusted BB84/DPS and separate BBM92 operation. BBM92 always requires
simultaneous visibility and recalculates singles and accidental coincidences on both arms.
Hardware, source, clock resources and security assumptions differ across protocols.
